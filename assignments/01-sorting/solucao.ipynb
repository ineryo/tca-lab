```python
import numpy as np

from tca.algorithms.sorting import selection_sort
```

```python
values_python = np.array(
    [5, 2, 4, 8, 4, 2, 1],
    dtype=np.float64,
)

selection_sort(
    values_python,
    backend="python",
)

values_python
```

```python
values_cpp = np.array(
    [5, 2, 4, 8, 4, 2, 1],
    dtype=np.float64,
)

selection_sort(
    values_cpp,
    backend="cpp",
)

values_cpp
```

```python
np.array_equal(values_python, values_cpp)
```

```python
@pytest.mark.parametrize("backend", ["python", "cpp"])
@pytest.mark.parametrize("size", [0, 1, 2, 10, 100])
def test_selection_sort_random_inputs(backend, size):
    rng = np.random.default_rng(42)

    values = rng.integers(
        -100,
        101,
        size=size,
    ).astype(np.float64)

    expected = np.sort(values.copy())

    selection_sort(values, backend=backend)

    np.testing.assert_array_equal(values, expected)
```
